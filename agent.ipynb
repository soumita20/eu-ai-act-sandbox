## Problem Statement

Organizations building LLM-powered software need to assess potential EU AI Act risks before deployment, but this process is often manual, inconsistent, and difficult to audit. Developers may not know whether an application’s purpose, prompts, data use, or model behavior could involve prohibited practices, high-risk use cases, or transparency obligations.

The EU AI Act Guardrail and Compliance Sandbox is a local-first automated testing platform that accepts an LLM application’s context and prompts, maps them to structured EU AI Act controls, runs adversarial safety tests, and produces an evidence-backed compliance report.

The system classifies each application as:
- Prohibited
- High Risk / Human Review Required
- Pass with Findings

It is intended to support technical governance and pre-deployment testing, not replace legal counsel or formal regulatory certification.
